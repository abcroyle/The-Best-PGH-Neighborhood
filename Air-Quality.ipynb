# Pittsburgh Air Quality: Daily AQI Data

## Introduction
For my individual project, I will investigate air quality across Allegheny County, focusing on areas within Pittsburgh. I want to compare air quality measurements from different monitoring locations to determine which areas generally have cleaner air.

## Metric
My individual metric is average **Air Quality Index (AQI)** at each monitoring location. I will compare the average AQI readings from different locations to determine which areas have the best air quality. Lower AQI values indicate better air quality, while higher values indicate more pollution and greater potential health concerns.

## Dataset
I will use the [Allegheny County Air Quality dataset](https://data.wprdc.org/dataset/allegheny-county-air-quality/resource/4ab1e23f-3262-4bd3-adbf-f72f0119108b)  from the Western Pennsylvania Regional Data Center (WPRDC).

This dataset contains daily air quality measurements collected by Allegheny County Health Department monitoring stations. It includes information such as the date, monitoring site, pollutant being measured, and Air Quality Index value. I will use these measurements to compare air quality across different locations.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Data/Neighborhoods.csv')
neighborhoods = df['hood'].dropna().unique().tolist()
df = pd.read_csv("Data/AirQual.csv")
col = "site"
```

```python
lookup = {n.strip().lower(): n for n in neighborhoods}

df["Neighborhood_clean"] = df[col].astype(str).str.strip().str.lower().map(lookup)
df[aqi_col] = pd.to_numeric(df[aqi_col], errors="coerce")
df = df.dropna(subset=["Neighborhood_clean", aqi_col])

avg_aqi = df.groupby("Neighborhood_clean")[aqi_col].mean()
avg_aqi = avg_aqi.reindex(neighborhoods)
```

```python
top5 = avg_aqi.sort_values().head(5).round(1).reset_index()
top5.columns = ["Neighborhood", "Avg AQI"]
top5.index = range(1, len(top5) + 1)
top5.index.name = "Rank"
top5
```
